```python
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model="openai/gpt-4o",
    base_url="https://openrouter.ai/api/v1",
    max_completion_tokens=512,
)

StreamingAgent = create_agent(
    model=model,
    system_prompt="You are a helpful Writer agent.",
    debug=True,
)
```

```python
StreamingAgent.invoke({"messages": [{"role": "user", "content": "Write a 100 words paragraph on artificial intelligence"}]})
```

```python
for message_chunk, metadata in StreamingAgent.stream(
    {"messages": [{"role": "user", "content": "Write a 200 words paragraph on artificial intelligence"}]},
    stream_mode="messages",
):
    if isinstance(message_chunk.text, str):
        print(message_chunk.text, end="|", flush=True)
```

Batch

Batching a collection of independent to Model to signaficalty improve performance and reduce cost, as the processing can be done in parllel.

```python
prompts = [
    {"messages": [{"role": "user", "content": "Why do parrots have colorful feathers?"}]},
    {"messages": [{"role": "user", "content": "How do airplanes fly?"}]},
    {"messages": [{"role": "user", "content": "What is quantum computing?"}]},
]
config ={
    'maxconcurrent': 5
}

responses = StreamingAgent.batch(prompts, config=config)

for response in responses:
    print(response["messages"][-1].content)
    
```
