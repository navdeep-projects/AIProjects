```python
from sklearn.tree import DecisionTreeClassifier
 
# Input data: [income, credit_score]
X = [
[30000, 550],
[40000, 600],
[60000, 700],
[80000, 750],
[100000, 800]
]
 
# Classification labels
y = [
"Rejected",
"Rejected",
"Approved",
"Approved",
"Approved"
]
 
# Create classification model
model = DecisionTreeClassifier()
 
# Train the model
model.fit(X, y)
 
# New application
new_application = [[70000, 720]]
 
# Predict
result = model.predict(new_application)
 
print(result[0])
```

```python
Variables

Variables are fundamental elements in programming used to store data that can be referenced and manipulated
in a program. In python, variables are created when you assign a value to them, and they do not need explicit
declaration to reserve memory space.  The declaration happens automatically when you assign a value to a variable.

Vedeo Outline:
- Introduction to Variables
- Variable Assignment
- Data Types
- Variable Naming Conventions
- Examples and Practice
- type checking and Conversion
- Dynamic Typing
- Practical Example and Common Erros
```

```python
a = 100
## Declaring and assigning variables
age = 32
height = 6.1
name = "Navdeep"
is_student = False
## Checking variable types
print("age:", age, "type:", type(age))
print("height:", height, "type:", type(height))
print("name:", name, "type:", type(name))
print("is_student:", is_student, "type:", type(is_student))
```

```python
## Type checking and conversionin Python
age=25

age_str = str(age)
print(age_str)
print("type of age_str:", type(age_str))    
age_int = int(age_str)
print(age_int)
print("type of age_int:", type(age_int))
```

```python
height = 6.1
print("height:", height, "type:", type(height))
height_int = int(height)
print("height_int:", height_int, "type:", type(height_int))
height_float = float(height_int)
print("height_float:", height_float, "type:", type(height_float))
```
