Importing Modules in phython: Modules and packakges

In python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
math.sqrt(25)
```

```python
from math import sqrt, pi
print(sqrt(16))
print(pi)
print(sqrt(25))
```

```python
import numpy as np

np.array([1, 2, 3, 4])
```

```python
from math import *
print(sqrt(16))
print(pi)
print(sqrt(25))
```

```python
from package.maths import addition, subtraction
print(addition(5, 3))
print(subtraction(5, 3))
```

```python
from package import maths
print(maths.addition(5, 3))
```
